- To edit code, simply select it and start typing.

## Other commends
Congratulations, you just ran your first jupyter code cell. Some other useful commands are:
- **ctrl s** saves the state of your notebook, allowing you to exit and return to your current session.
- The "Kernel" menu controls the python engine that is running your code:
  - **Restart kernel** kills the kernel and starts it again. It does not clear any outputs from the code cells you've run, but it *will* lose the values of any variables that had been set, so is not recommended.
  - **Restart kernel and clear all cells** both restarts the kernel and clears all the outputs, so it will be obvious that you are starting again. This is the recommended command to use if you run into trouble and need to restart.
## Gotchas
1. It is important to run all python cells as they are often dependent on each other: don't skip any!
2. Occasionally you may find the page isn't displaying correctly, e.g. it is blank from the last run cell down. Refreshing your browser tab (**ctrl r**) should fix this.
3.  Sometimes a code cell can't be selected. If this happens, go back to one you can select, then try again.
