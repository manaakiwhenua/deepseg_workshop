# A brief introduction to deep learning

In this section we'll learn what "deep learning" actually is. To be precise, we are learning about ***deep convolutional networks for imagery***; this is a subset of the broader subject of deep learning.

Before we can dive into deep learning, we need to back up a step: deep convolutional networks are a kind of ***artificial neural network***, which is itself an example of a ***machine learning*** model, so we'll take a quick look at those first.

## Machine Learning models
<center><img src="images/ml.png" width=500/></center>

A machine learning (ML) model is a model that can be trained to perform some kind of task. ML models come in many flavours, and can be applied to a a wide range of problem domains:
- Some models perform ***classification*** by selecting from a set of discrete outcomes, also known as labelling (e.g. the species of this animal is "possum")
- Other models perform ***regression***, i.e. they generate a continuous (numeric) output, such as predicting the rain that will fall next year in mm
- ML models can work with a wide variety of datatypes, including tables of categorical and numeric values, sound clips, imagery, video, etc

ML models are generally "trained" by presenting them with pairs of (input, desired output): this is known as ***supervised learning***. Other types of training also exist, including ***unsupervised learning*** (the desired outcome is not known, e.g. clustering), and ***reinforcement learning*** (the model learns to improve on its past performance). All the examples we will look at are ***supervised learning***.

## Artificial neural networks
<center><img src="images/neuron.png" width=500/></center>

An artificial neural network (or ANN) is a machine learning model that is conceptually based (a gross simplification of) the human brain. Our brains are known to be full of ***neurons*** which accept input from either the outside world (e.g. through our eyes) or other neurons via connections called ***synapses***. The neurons process this incoming information and generate outputs that in turn are connected via synapses to other neurons, or to the outside world (e.g. by connecting to muscles that allow us to take action).

ANNs operate on a similar basis: 
- Each artifical neuron (or ***perceptron***) accepts one or more inputs and processes them to produce an output. The processing has two steps:
  - The inputs are summed
  - The result is fed into an activation function that converts the continuous sum value into (approximately) a 0 (not activated) or 1 (activated)
- The neurons are arranged in ***layers***:
  - The ***input layer*** accepts inputs from the outside world (i.e. the input data) and outputs the results to the next (hidden) layer
  - ***Hidden layers*** accept inputs from other layers and output the results to either another hidden layer or the output layer
  - The ***output layer*** accepts inputs from a hidden layer and generates the final result (typically the class, e.g. "possum")
- The neurons are ***fully connected***: Every neuron is connected to every output from the previous layer. (The input layer is connected to every input value of the incoming data).
  - The neurons are connected via ***weighted connections***, where the weights indicate the contribution of this particular input to the final ooutput value; it is these weights that are learned during training.
 
<center><img src="images/neural_network.png" width=500/></center>


Like all supervised learning, Neural networks are trained by feeding them examples of (input, output). For example, to train a model to predict whether a patient will develop diabetes, we would pass it many examples of (patient history, prognosis). The training cycle is:
1. Initialise the weights in the network to small random values
2. Pass in a batch of examples (called a "***minibatch***"), and let the network predict the outputs
3. Compare the predicted and actual outputs: the difference is the error (or ***loss***) that we will use to update the network weights
4. Update the weights via ***backpropagation***:
  - compute the contribution from each neuron in the output layer to the error
  - update the output weights for the each neron in the output layer in proportion to ***loss*** X ***output*** X ***learning rate*** where ***learning rate*** is a (small) constant that controls how fast the weights change.
5. Repeat for each layer, working backward to the input layer
6. Repeat 2-5 many times: each complete pass of all of the training data is called an ***epoch***.

And that's all there is to it! By training the network we are optimising the parameters of a very large non-linear equation that transforms the input data into the corresponding output class.

## ANNs and images: convolutional neural networks

In theory, we could use the above fully connected feedforward network to learn to classify images by simply connecting the input neurons to every pixel in the image, and this in fact was done. Unfortunately this approach has two major drawbacks:
1. The computational processing required to train such a network is massive, and only feasible for small image sizes and/or simple networks
2. The resulting network would have poor ability to generalise, because it would learn the exact location and size of each object it was trained on.

Convolutional neural networks (or ***CNNs***) overcome both of the problems. Instead of processing all pixels in an image at once, neurons are arranged into small 2D matrices (e.g. 3x3) that act as small ***filters***.

<center><img src="images/convolutional_filter.png" width=500/></center>


1. This filter is scanned, or ***convolved*** over the image, generating a new image or ***feature map*** where each pixel now represents the strength with which the filter matched to this area in the image.
2. Each layer in the network contains multiple filters, and therefore generates multiple feature maps from the image.
3. Hidden layers now take as input the feature maps from the previous layer and generate new feature maps.
4. Between each layer, the feature maps are downscaled (typically by half); this has two advantages:
   - The image is filtered at multiple resolutions, making the process more size-invariant
   - At each downscaling, detail about the image is lost. This forces the network to find patterns that are increasingly abstract.

<center><img src="images/cnn.png" width=500/></center>

The output of a convolutional network is no longer an answer such as "possum" but a set of feature maps that caputre abstract patterns that are present in teh image, and where they are found. In other words, the convolutional neural network is an automatic feature extractor, where those features are a highly compressed representation of the original image (a bit like ***principal component analysis*** for numeric data). These features can now be fed into our favourite ML model to generate the answer.

## We need to go deeper: deep convolutional networks or deep learning
CNNs greatly increase the utility of neural networks for image processing, but they are still computationally expensive which limits how deep they can go (i.e. how many layers they can have) and therefore the level of abstraction they can achieve - vanilla CNNs such as ***AlexNet*** consisted of fewer than 10 convolutional layers. To overcome this, many architectural tricks have been devised to enable much deeper networks to be built. Based on an idea called "***networks within networks***", modern deep learning networks consist of many smaller networks chained together to give hundreds (or even thousands) of layers and many millions (or billions) of parameters. These extremely powerful models can discriminate between thousands or millions of discrete classes. For example, the diagram below is the high-level architecture of the ***Xception*** model, which you will encounter shortly. This medium-sized model consists of 81 layers and has around 22 million weights (trainable parameters).

## Practical deep learning: the tensorflow/keras technology stack
<center><img src="images/tensorflow.png" width=500/></center>

Finally, we need some tools to build, train and run our deep learning networks. Whilst we could develop all the code from scratch, it's far more likely we will use an existing framework such as ***Tensorflow/keras*** or ***Pytorch***. For this tutorial we will be using the Tensorflow/keras framework. This provides us with the following key features:
- A simple python interface (Keras) for developing our models
- A library of pre-trained models saving us potentially thousands of hours of processing time
- Ability to run our code on a variety of hardware platforms, most importantly including a general-purpose graphics processing unit (GPGPU or just GPU for short) which can greatly speed up processing time.
